КОНТЕКСТ: Я вчуся аналізу даних. У таблиці студентів стовпець salary_usd = 0 означає «не працевлаштований», а не справжню зарплату. Я хочу середню і медіану лише серед працевлаштованих і пояснення, чому нулі не можна залишати.

ПРИКЛАД ДАНИХ (вигадані):
student_id,status,salary_usd
S1,Placed,70000
S2,Not Placed,0
S3,Placed,90000
S4,Not Placed,0
S5,Placed,80000
S6,Placed,100000

ФОРМАТ ВІДПОВІДІ: 1) 3 короткі пункти пояснення; 2) один блок коду pandas; 3) очікувані числа для цього прикладу.

```python
from io import StringIO

csv = """student_id,status,salary_usd
S1,Placed,70000
S2,Not Placed,0
S3,Placed,90000
S4,Not Placed,0
S5,Placed,80000
S6,Placed,100000"""

df_ai = pd.read_csv(StringIO(csv))

# Неправильно: нулі входять у розрахунок
naive_mean = df_ai["salary_usd"].mean()
naive_median = df_ai["salary_usd"].median()

# Правильно: лише працевлаштовані
placed = df_ai[df_ai["status"] == "Placed"]
mean_salary = placed["salary_usd"].mean()
median_salary = placed["salary_usd"].median()

# Частка працевлаштованих
placement_rate = (df_ai["status"] == "Placed").mean()

print(f"Наївне середнє:   {naive_mean:.2f}")
print(f"Наївна медіана:   {naive_median:.2f}")
print(f"Середнє (Placed): {mean_salary:.2f}")
print(f"Медіана (Placed): {median_salary:.2f}")
print(f"Частка Placed:    {placement_rate:.1%}")

# Альтернатива: замінити 0 на NaN, pandas ігнорує NaN автоматично
df_ai["salary_clean"] = df_ai["salary_usd"].mask(df_ai["salary_usd"] == 0)
print(df_ai["salary_clean"].astype("float").mean())
```
